```python
import yfinance as yf
import plotly.graph_objects as go
```

```python
ticker=yf.Ticker('INFY')
hist_data = ticker.history(period="6mo",interval="1d")
hist_data=hist_data[['Open','High','Low','Close','Volume']]
hist_data
```

```python
fig = go.Figure(data=[go.Candlestick(x=hist_data.index, open=hist_data["Open"], high=hist_data["High"], low=hist_data["Low"], close=hist_data["Close"])])
fig.update_layout(title="Last 6 months Candlestick chart of INFOSYS")
fig.show()
```

**On July 23rd a strong bearish hammer is formed leads to fall of stock till Aug 11th**


**On Aug 11th there is a reverse hammer followed with bullish candles is sign of bullish nature which continued till Aug 22nd.**

**Then an vening star pattern formed on Aug 25th shows bearish nature.**

**Again on Oct 16th hammer pattern at the support level leads to bullish**

**And then on Oct 23rd Doji followed by Bearish Harami leads to bearish nature till Nov 6th.**
